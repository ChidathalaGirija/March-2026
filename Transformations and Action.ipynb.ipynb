```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
spark = SparkSession.builder.appName("RetailTransformation").getOrCreate()
```

```python
# Retail Dataset
data =[
    (1,"2026-01-05","Bangalore","Laptop","Electronics",2,55000),
    (2,"2026-01-10","Hyderabad","Mobile","Electronics",2,25000),
    (3,"2026-02-20","Chennai","Sheos","Fashion",5,3000),
    (4,"2026-02-20","Banglore","Tv","Electronics",1,45000),
    (5,"2026-03-01","Pune","Shirt","Fashion",4,1200),
    (6,"2026-03-12","Hyderabad","Rice Bag","Grocery",10,900),
    (7,"2026-04-05","Chennai","Headphones","Electronics",6,2000),
    (8,"2026-04-18","Pune","Jeans","Fashion",3,2500),
    (9,"2026-05-08","Banglore","Oil Packet","Grocery",8,180),
    (10,"2026-05-20","Hyderabad","Watch","Fashion",2,5000)
     
]
```

```python
columns =[ 
"order_id",
 "order_date",
 "city",
 "product",
 "category",
 "quantity",
 "price"
 ]
df=spark.createDataFrame(data,columns)
 # Convert string date to date type
df =df.withColumn("order_date",to_date(col("order_date")))
df.show()
df.printSchema()
```

```python
df_filter = df.filter(df.category == "Electronics")
df_filter.show()
```

```python

df_amount = df.withColumn("total_amount",col("quantity")*col("price"))
df_amount.show()
```

```python
# excract year,month,day from date
df_date = df.withColumn("year",year(col("order_date")))\
    .withColumn("month",month(col("order_date")))\
        .withColumn("day",dayofmonth(col("order_date")))

df_date.show()
```

```python
# filter date range
df_date_filter = df.filter(
    (col("order_date")>= "2026-02-01")&
    (col("order_date")<="2026-04-30")
)
df_date_filter.show()
```

```python
df.show()
```

```python
df_category_sales = df.withColumn("total_amount",col("quantity")*col("price"))\
    .withColumn("month",month(col("order_date")))\
        .groupBy("month")\
            .agg(sum("total_amount").alias("total_sales"))

df_category_sales.show()
```

```python
df_drop = df.drop("price")
df_drop.show()
```

```python
#withcolumnRenamed transformation
df_renamed = df.withColumnRenamed("city","stored_city")
df_renamed.show()
```

```python
# distinct transformation
df_distinct_city = df.select("city").distinct()
df_distinct_city.show()
```

```python
# sort transformation
df_sort = df.sort(col("order_date").desc())
df_sort.show()
```

```python
# limit transformation
df_limit = df.limit(5)
df_limit.show()
```

```python
# dropduplicates transformation
df_no_duplicate = df.dropDuplicates(["order_id"])
df_no_duplicate.show()
```
