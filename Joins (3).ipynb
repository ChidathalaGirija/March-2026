```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("joins").getOrCreate()
```

```python
 # Employee Data
emp_data = [
    (1,"Girija",101),
    (2,"Nikhil",102),
    (3,"Rupa Rani",103),
    (4,"Divya",104),
    (5,"Rani",None)
 ]
emp_df = spark.createDataFrame(emp_data,["emp_id","emp_name","dept_id"])
 # Department Data
dept_data = [
    (101,"HR"),
    (102,"IT"),
    (103,"Finance"),
    (104,"Sales")
 ]
 dept_df = spark.createDataFrame(dept_data,["dept_id","dapt_name"])
 emp_df.show()
 dept_df.show()
```

```python
df_inner = emp_df.join(dept_df,"dept_id",'inner')
df_inner.show()
```

```python
df_left = emp_df.join(dept_df,"dept_id","left")
df_left.show()
```

```python
df_right = emp_df.join(dept_df,"dept_id","right")
df_right.show()
```

```python
df_full= emp_df.join(dept_df,"dept_id","full")
df_full.show()
```

```python
df_semi = emp_df.join(dept_df,"dept_id","left_semi")
df_semi.show()
```

```python
df_anti = emp_df.join(dept_df,"dept_id","left_anti")
df_anti.show()
```

```python
df_cross = emp_df.crossJoin(dept_df)
df_cross.show()
```
